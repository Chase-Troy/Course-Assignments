ConvNet Experiments – MNIST vs Fashion-MNIST

```python
# Import basic libraries
import time
import tensorflow as tf

# Import datasets
from tensorflow.keras.datasets import mnist, fashion_mnist

# Import tools for building the neural network
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense

# For converting labels to categorical form
from tensorflow.keras.utils import to_categorical
```

Preparing and Normalizing the Data

```python
def prepare_data(x_train, y_train, x_test, y_test):

    # reshape images so that they include the channel dimension
    x_train = x_train.reshape((x_train.shape[0], 28, 28, 1))
    x_test = x_test.reshape((x_test.shape[0], 28, 28, 1))

    # convert pixel values from 0–255 to 0–1
    x_train = x_train.astype("float32") / 255
    x_test = x_test.astype("float32") / 255

    # convert labels into categorical format
    y_train = to_categorical(y_train)
    y_test = to_categorical(y_test)

    return x_train, y_train, x_test, y_test
```

Building the Original ConvNet Model

```python
def build_original_model():

    model = Sequential()

    # first convolution layer
    model.add(Conv2D(32, (3,3), activation='relu', input_shape=(28,28,1)))
    model.add(MaxPooling2D((2,2)))

    # second convolution layer
    model.add(Conv2D(64, (3,3), activation='relu'))
    model.add(MaxPooling2D((2,2)))

    # convert feature maps to a single vector
    model.add(Flatten())

    # fully connected layers
    model.add(Dense(128, activation='relu'))
    model.add(Dense(64, activation='relu'))

    # output layer for 10 classes
    model.add(Dense(10, activation='softmax'))

    model.compile(
        optimizer='adam',
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )

    return model
```

Function to Train and Evaluate the Model

```python
def train_model(model, x_train, y_train, x_test, y_test):

    print("\nTraining the model...\n")

    start_time = time.time()

    model.fit(
        x_train,
        y_train,
        epochs=5,
        batch_size=128,
        verbose=1
    )

    end_time = time.time()

    # evaluate model performance
    loss, accuracy = model.evaluate(x_test, y_test)

    print("\nTest Accuracy:", accuracy)
    print("Training Time:", round(end_time - start_time, 2), "seconds")
```

Original Model with MNIST

```python
print("Experiment 1: Training with MNIST dataset")

# load dataset
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# preprocess the data
x_train, y_train, x_test, y_test = prepare_data(
    x_train, y_train, x_test, y_test
)

# build and train the model
model = build_original_model()

train_model(model, x_train, y_train, x_test, y_test)
```

Original Model with Fashion-MNIST

```python
print("Experiment 2: Training with Fashion-MNIST dataset")

# load Fashion-MNIST
(x_train, y_train), (x_test, y_test) = fashion_mnist.load_data()

# preprocess the data
x_train, y_train, x_test, y_test = prepare_data(
    x_train, y_train, x_test, y_test
)

# build and train the same model
model = build_original_model()

train_model(model, x_train, y_train, x_test, y_test)
```

Results and Discussion

The convolutional neural network model was trained on two datasets: MNIST and Fashion-MNIST. The goal was to compare the model’s performance and training time on both datasets.

Results

MNIST Dataset

  Test Accuracy: 0.9895 (≈ 98.95%)

  Training Time: 337.21 seconds

Fashion-MNIST Dataset

Discussion

The model performed better on the MNIST dataset, achieving an accuracy of approximately 98.95%, while the accuracy on the Fashion-MNIST dataset was about 90.07%.

This difference occurs because MNIST images are simpler. They contain handwritten digits with clear shapes and less variation. In contrast, Fashion-MNIST images represent clothing items, which are more complex and visually similar to one another. For example, items such as shirts, coats, and pullovers may have overlapping features, making classification more challenging.

In terms of training time, the Fashion-MNIST model took slightly longer to train (about 411 seconds) compared to the MNIST model (about 337 seconds). Although both datasets have the same number of images and the same image size (28×28 pixels), the increased complexity of Fashion-MNIST images may cause the model to require more computation during training.

Overall, the results show that dataset complexity has a significant impact on model accuracy and training performance.

Model Without the First Dense Layer

```python
def build_model_without_dense():

    model = Sequential()

    model.add(Conv2D(32,(3,3),activation='relu',input_shape=(28,28,1)))
    model.add(MaxPooling2D((2,2)))

    model.add(Conv2D(64,(3,3),activation='relu'))
    model.add(MaxPooling2D((2,2)))

    model.add(Flatten())

    # only one dense layer now
    model.add(Dense(64,activation='relu'))

    model.add(Dense(10,activation='softmax'))

    model.compile(
        optimizer='adam',
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )

    return model
```

Experiment 3 – Removing the First Dense Layer

```python
print("Experiment 3: Removing the first Dense layer")

(x_train, y_train), (x_test, y_test) = mnist.load_data()

x_train, y_train, x_test, y_test = prepare_data(
    x_train, y_train, x_test, y_test
)

model = build_model_without_dense()

train_model(model, x_train, y_train, x_test, y_test)
```

Model with an Extra Dense Layer (4096 Neurons)

```python
def build_large_model():

    model = Sequential()

    model.add(Conv2D(32,(3,3),activation='relu',input_shape=(28,28,1)))
    model.add(MaxPooling2D((2,2)))

    model.add(Conv2D(64,(3,3),activation='relu'))
    model.add(MaxPooling2D((2,2)))

    model.add(Flatten())

    # large dense layer similar to pretrained convnets
    model.add(Dense(4096,activation='relu'))

    model.add(Dense(128,activation='relu'))
    model.add(Dense(64,activation='relu'))

    model.add(Dense(10,activation='softmax'))

    model.compile(
        optimizer='adam',
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )

    return model
```

Experiment 4 – Adding Dense(4096)

```python
print("Experiment 4: Adding a Dense layer with 4096 neurons")

(x_train, y_train), (x_test, y_test) = mnist.load_data()

x_train, y_train, x_test, y_test = prepare_data(
    x_train, y_train, x_test, y_test
)

model = build_large_model()

train_model(model, x_train, y_train, x_test, y_test)
```

Results and Discussion

In this exercise, the ConvNet model was modified in two ways. First, the first Dense layer was removed. Second, a large Dense layer with 4096 neurons was added before the existing Dense layers. The goal was to observe how these changes affect prediction accuracy and training time.

Experiment 3 – Removing the First Dense Layer

Results:

Test Accuracy: 0.9894 (≈ 98.94%)

Training Time: 336.21 seconds

Observation:

Removing the first Dense layer did not significantly reduce the model’s accuracy. The model still achieved an accuracy close to 98.94%, which is almost the same as the original model. This indicates that the removed Dense layer was not contributing significantly to the final predictions.

However, the model became slightly simpler, which helps reduce the number of parameters and makes the network easier to train. The training time remained approximately the same as the original model because the change in parameters was relatively small.

Experiment 4 – Adding a Dense Layer with 4096 Neurons

Results:

Test Accuracy: 0.9912 (≈ 99.12%)

Training Time: 901.83 seconds

Observation:

Adding a large Dense layer with 4096 neurons slightly improved the model’s accuracy, increasing it to about 99.12%. This improvement occurs because the larger Dense layer increases the model’s capacity to learn complex patterns in the data.

However, this modification greatly increased the training time, rising from about 336 seconds to over 900 seconds. This happens because the 4096-neuron layer introduces millions of additional parameters, which significantly increases the amount of computation required during training.

Conclusion

The experiment shows that increasing the size of Dense layers can slightly improve prediction accuracy, but it comes at the cost of much longer training times and higher computational requirements. In contrast, removing a Dense layer can simplify the model with little to no loss in performance.

This demonstrates the importance of balancing model complexity, accuracy, and computational efficiency when designing neural networks.
