```
import kagglehub
import pandas as pd
import os

# Dataset download
path = kagglehub.dataset_download("lovishbansal123/diamond-dataset")
print("Path to dataset files:", path)

# Showing files in folder
os.listdir(path)
```

```
# Loading dataset into Pandas
df = pd.read_csv(os.path.join(path, "diamonds.csv"), index_col=0)

# Displaying the first seven rows of the DataFrame
df.head(7)
```

```
# Display the last seven rows of the DataFrame
df.tail(7)
```

```
# DataFrame method describe for statistics
numeric_cols = ["carat", "depth", "table", "price", "x", "y", "z"]
df[numeric_cols].describe()
```

```
# Series method describe statistics all together
df[["cut", "color", "clarity"]].describe()
```

```
# Series method unique catagory values
for col in ["cut", "color", "clarity"]:
    print(col, "->", df[col].unique())
```
